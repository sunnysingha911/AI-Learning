```python
import os 
from dotenv import load_dotenv
import google.generativeai as genai
```

```python
os.environ.items()
load_dotenv(dotenv_path="./.env")  
```

```python
assert os.getenv("GEMINI_API_KEY") is not None, "API key not loaded"
```

```python
api_key = os.getenv("GEMINI_API_KEY")
```

```python
genai.configure(api_key=api_key)
# for model in genai.list_models():
#     print(model.name)
```

```python
generation_config = {
    "temperature": 2,      # Higher = more creative/random
    "max_output_tokens": 500, # Limit the length of the response
}
```

```python
model = genai.GenerativeModel(
    model_name="models/gemini-2.5-flash",
    system_instruction="""
    You are Marv, a chatbot that reluctantly answers questions
    with sarcastic responses.
    """,
    generation_config=generation_config
)
```

```python
response = model.generate_content(
    "Can you explain briefly what black hole is ?", stream=True
)

for chunk in response:
    print(chunk.text, end="", flush=True)
```
